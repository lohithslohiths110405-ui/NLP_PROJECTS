```python
# Import required libraries
import pandas as pd
import re
import string
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.sentiment import SentimentIntensityAnalyzer
```

```python
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("vader_lexicon")
```

```python
# Load dataset
df = pd.read_csv(r"C:\Users\Admin\Desktop\NLP\Hotel_Reviews.csv")

# Display first 5 rows
print(df.head())
```

```python
print(df.shape)

print(df.columns)
```

```python
# Keep only required columns
df = df[["Hotel_Name",
         "Positive_Review",
         "Negative_Review",
         "Reviewer_Score"]]
```

```python
# Combine both reviews into one column
df["Review"] = df["Positive_Review"] + " " + df["Negative_Review"]

# Display first few rows
print(df.head())
```

```python
print(df.isnull().sum())

# Remove missing values
df.dropna(inplace=True)
```

```python
# English stopwords
stop_words = set(stopwords.words("english"))

# Create lemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
# Function to clean text
def clean_text(text):

    # Convert to lowercase
    text = text.lower()

    # Remove numbers
    text = re.sub(r"\d+", "", text)

    # Remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))

    # Split into words
    words = text.split()

    cleaned_words = []

    # Remove stopwords
    for word in words:

        if word not in stop_words:

            word = lemmatizer.lemmatize(word)

            cleaned_words.append(word)

    return " ".join(cleaned_words)
```

```python
# Apply cleaning function
df["Clean_Review"] = df["Review"].apply(clean_text)

print(df[["Review","Clean_Review"]].head())
```

```python
# Create sentiment analyzer
sia = SentimentIntensityAnalyzer()
```

```python
# Function to predict sentiment
def get_sentiment(review):

    score = sia.polarity_scores(review)

    compound = score["compound"]

    if compound >= 0.05:
        return "Positive"

    elif compound <= -0.05:
        return "Negative"

    else:
        return "Neutral"
```

```python
# Predict sentiment
df["Sentiment"] = df["Clean_Review"].apply(get_sentiment)

print(df[["Review","Sentiment"]].head())
```

```python
# Count sentiments
print(df["Sentiment"].value_counts())
```

```python
# Total reviews
total_reviews = len(df)

# Positive reviews
positive_reviews = len(df[df["Sentiment"]=="Positive"])

# Satisfaction percentage
satisfaction = (positive_reviews/total_reviews)*100

print("Customer Satisfaction =",round(satisfaction,2),"%")
```
