```python
# Import pandas for handling data
import pandas as pd

# Import regular expressions
import re

# Import string module to remove punctuation
import string

# Natural Language Processing library
import nltk

# Stopwords are common words like "the", "is", "and"
from nltk.corpus import stopwords

# Used to convert words into their base form
from nltk.stem import WordNetLemmatizer

# Split dataset into training and testing
from sklearn.model_selection import train_test_split

# Convert text into numerical features
from sklearn.feature_extraction.text import TfidfVectorizer

# Machine Learning algorithm
from sklearn.naive_bayes import MultinomialNB

# Evaluation metrics
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
# Download stopwords
nltk.download("stopwords")

# Download WordNet for lemmatization
nltk.download("wordnet")
```

```python
# Read the JSON dataset
df = pd.read_json(r"C:\Users\Admin\Desktop\NLP\News_Category_Dataset_v3.json", lines=True)

# Display first 5 rows
df.head()
```

```python
# Display number of rows and columns
print(df.shape)

# Display column names
print(df.columns)
```

```python
# Keep only category, headline and short description
df = df[["category", "headline", "short_description"]]
```

```python
# Create one text column by joining headline and description
df["text"] = df["headline"] + " " + df["short_description"]

# Keep only text and category columns
df = df[["text", "category"]]

# Display first few rows
df.head()
```

```python
# Categories needed for the project
categories = [
    "POLITICS",
    "SPORTS",
    "BUSINESS",
    "TECH",
    "ENTERTAINMENT"
]

# Filter the dataset
df = df[df["category"].isin(categories)]

# Check available categories
print(df["category"].value_counts())
```

```python
# Check missing values
print(df.isnull().sum())

# Remove missing values
df.dropna(inplace=True)
```

```python
# Store English stopwords
stop_words = set(stopwords.words("english"))

# Create lemmatizer object
lemmatizer = WordNetLemmatizer()
```

```python
# Function to clean text
def clean_text(text):

    # Convert to lowercase
    text = text.lower()

    # Remove numbers
    text = re.sub(r"\d+", "", text)

    # Remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))

    # Split sentence into words
    words = text.split()

    # Remove stopwords and apply lemmatization
    cleaned_words = []

    for word in words:
        if word not in stop_words:
            cleaned_word = lemmatizer.lemmatize(word)
            cleaned_words.append(cleaned_word)

    # Join words back into a sentence
    return " ".join(cleaned_words)
```

```python
# Apply cleaning function to every article
df["clean_text"] = df["text"].apply(clean_text)

# Display cleaned data
df.head()
```

```python
# Create TF-IDF object
vectorizer = TfidfVectorizer(max_features=5000)

# Convert text into numerical features
X = vectorizer.fit_transform(df["clean_text"])

# Target variable
y = df["category"]
```

```python
# Split data into training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
# Create Naive Bayes model
model = MultinomialNB()

# Train the model
model.fit(X_train, y_train)
```

```python
# Predict categories for testing data
y_pred = model.predict(X_test)
```

```python
# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
# Display precision, recall and F1-score
print(classification_report(y_test, y_pred))
```

```python
# Display confusion matrix
cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
# Enter your own news article
news = [
    "India wins the cricket world cup after defeating Australia in the final."
]

# Clean the text
news_clean = [clean_text(news[0])]

# Convert into TF-IDF features
news_vector = vectorizer.transform(news_clean)

# Predict category
prediction = model.predict(news_vector)

print("Predicted Category:", prediction[0])
```
